
## Strategy toolbox: levers you can add to your prompt

Your prompt is not just a task description; it also tells the assistant *how to work*. Pick two or three levers below, paste or adapt the sentence into the **Working approach** line of your prompt, and see what changes. Four prompts that pull different levers were run on this same task; see [what changed in each](examples/prompt_iteration/README.md).

**Change one thing at a time if you want to know what helped**, and keep your evaluation fixed: same splits, validation for choices, test only at the end. A longer prompt is not automatically a better one.

| Lever | Add to your prompt | Watch out for |
| --- | --- | --- |
| Hypothesis first | "My hypothesis is ___ because ___. Test it against a baseline on validation data and report whether it helped." | A vague hypothesis gives a vague experiment. |
| Anchor and frame | "Follow the ten steps of 01_Ten_Steps_in_Prompts.ipynb. Act as a chemist and ML practitioner writing for a mixed-experience audience." | The assistant may copy the template's structure without thinking about your data. |
| Plan before code | "First write a short plan: outline, decisions you expect to make, artifacts to save. Then proceed." | A plan can be wrong; read it before the code runs. |
| Small verified steps | "Build one step at a time. Run it and check the output before building on it." | Slower and chattier; best when you want to review as you go. |
| Name the guardrails | "Molecule-disjoint splits, preprocessing fitted on training data only, validation for all choices, test set opened once." | Only the rules you name are enforced; check that they were. |
| Ask for specific checks | "Add assertions for splits and units. Reload the saved model in a fresh process. Warn when a molecule is unlike the training data." | Each request adds code and runtime. |
| Self-consistency | "At key decisions (cleaning rule, features, final model) try more than one defensible choice and keep what the evidence agrees on." | More experiments to read; agreement is not proof of correctness. |
| Critique and revise | "After the first complete run, review it for leakage, unsupported claims and weak spots, then fix what you find." | The assistant can approve its own work; you still review it. |

Whatever you choose, inspect the real outputs. An assistant can describe a check it never ran.
