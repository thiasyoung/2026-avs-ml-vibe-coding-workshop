THE TEN STEPS

Step 1: Set Goal

Briefly state the already-defined goal: predict the dataset's reported boiling point from SMILES. Use MAE in kelvin as the primary selection metric, with RMSE and R-squared as supporting metrics. Set a transparent workshop success criterion before fitting, such as improving held-out MAE over a training-median baseline. Do not invent an industrial accuracy requirement or promise that the criterion will be met. Explain the intended use and relevant limits of the available labels. Do not claim verified normal boiling points at a fixed pressure if the retained provenance does not establish that condition.

Step 2: Define Inputs and Outputs

Briefly define SMILES as the inference input and a numerical boiling-point prediction in K as the output. Celsius may also be displayed with an explicit conversion. Specify how invalid, unsupported, and unfamiliar molecules will be handled. No target value or external lookup should be needed to predict a new SMILES.

Step 3: Collect Data

State that collection is complete. Identify the selected local file, its source and units, and any provenance limitations. Do not rerun acquisition. Keep Steps 1-3 short; devote the implementation and teaching detail to Steps 4-10.

Step 4: Examine Data

Load and preview the actual data. Audit the schema, missing and nonnumeric targets, nonfinite values, invalid or empty SMILES, duplicate rows, equivalent molecular structures, conflicting repeated measurements, and unusual chemistry or temperatures. Inspect mixtures, salts, charges, and stereochemistry where relevant. Report counts and representative flagged examples with reasons; a suspicious observation is not automatically an error.

Use source evidence to establish units, not the apparent temperature range alone. Distinguish deterministic validity checks from exploratory decisions that could influence the model. After the minimal validity and identity checks needed to define eligible molecules, reserve the data splits before target-informed exploration or learned preprocessing. Use training data for plots and investigations that guide modeling. Include one or two readable data figures with units and sample counts.

Step 5: Data Carpentry

Implement and explain a conservative cleaning policy, retaining an audit of exclusions and changes. Preserve original SMILES and row identifiers for traceability. Canonicalize molecular identity consistently, keep equivalent structures in one split, and handle repeated measurements with a documented rule. Do not silently merge chemically distinct forms or remove difficult examples because they hurt performance. Define the supported chemistry and apply the same policy at inference.

Choose and freeze a reproducible train/validation/test design appropriate to the intended use. An 80/10/10 split of unique molecules with seed 42 is a reasonable workshop default. Explain what a random molecular split estimates and its limits for unfamiliar chemical families. If you choose scaffold or similarity grouping, justify it and handle acyclic molecules sensibly. Save assignments and assert that molecular identities and any required groups do not overlap. Revisit the design in Step 7 without creating new splits to improve scores.

Engineer a compact, chemically motivated representation from SMILES. Consider how size, branching, functional groups, intermolecular interactions, or molecular fingerprints could help; select what fits the data and runtime. Explain the proposed features as hypotheses rather than guaranteed physical laws. Fit imputation, scaling, feature selection, and other learned transformations only on training data, and within each training fold if cross-validation is used. Save a clean working table and the representation specification. Test the usefulness of feature choices in Step 7.

Step 6: Algorithm

Build a training-median dummy baseline and two complementary, affordable regression candidates. Choose the candidates yourself based on dataset size, representation, nonlinear chemistry, interpretability, and runtime. Do not default to a neural network because neighboring notebooks use one. Keep a consistent fit/predict interface and explain the tradeoffs. Use reusable pipelines so training and inference apply identical transformations. For iterative models, use validation-based stopping and restore the selected checkpoint; for other models, do not invent epoch histories.

Step 7: Experiment

Run a small, deliberate set of fair experiments on the frozen training and validation partitions. Compare model families, perform a bounded settings search, and include at least one controlled feature or representation ablation. Reuse the same partitions or folds so comparisons are meaningful. Record the hypothesis, settings, feature set, seed, runtime, train/validation metrics, and predictions for each numbered experiment. Save metadata.json, predictions.csv, training_history.csv when applicable, and an experiment_summary.csv. Include a clear validation-comparison figure. Do not calculate or inspect test predictions or performance during this step.

Use chemical and machine-learning knowledge creatively: propose two plausible ways to make the predictions more useful, then implement and evaluate one well-motivated enhancement. Examples include a chemistry-informed descriptor/fingerprint combination, a simple physical trend with a learned residual correction, an explanation using similar training molecules, or a defensible applicability or uncertainty diagnostic. These are suggestions, not a mandatory method list. Define the benefit and evaluate it against a simpler control using validation evidence. An honest finding that the idea did not help is acceptable.

If you report prediction intervals, use an appropriate calibration design kept separate from model selection, document its assumptions, and later report empirical coverage and interval width. Model disagreement and chemical similarity alone are not calibrated probabilities or guaranteed error bounds.

Step 8: Analysis

Analyze the validation experiments, identify overfitting or weak chemical subgroups, and select the most promising configuration by the declared criterion. Include chemical interpretation of representative successes and failures and acknowledge small subgroup counts. Describe feature importance or related explanations as model associations, not proof of causation.

Keep the test set sealed here because Step 9 still allows optimization. Explain this deliberate refinement of the template. Prepare the final evaluation/reporting code, but run it only after all modeling decisions are frozen.

Step 9: Optimization

Make this step substantive: carry out one focused, affordable improvement motivated by the training/validation analysis. Change one main factor so its effect is understandable. Compare it with the previous best using the same validation procedure, record the result, and keep the simpler or better-supported model if the change does not help. Limit iteration; do not search until a desired score appears. Freeze the selected pipeline, preprocessing, support policy, and any calibration before opening the test set. Keep the evaluated model trained on its declared training partition for this teaching example, so validation and test reports remain easy to interpret.

Step 10: Solved?

Now evaluate the frozen selected model and the predefined dummy baseline on the held-out test set. Report train, validation, and test sample counts, MAE, RMSE, and R-squared, clearly identifying the test results as the final generalization estimate for this split. Produce readable parity and residual plots and a small set of representative molecular error examples. Include any final applicability/uncertainty results supported by the chosen method. These diagnostics must not trigger further tuning on this test set.

Answer the Step 1 success criterion using actual results, discuss limitations, and propose one useful next experiment. Failure to beat the baseline is a result to explain, not a reason to change the test split or fabricate improvement.

Finish with a working predict_boiling_point(smiles) helper supporting one SMILES or a list. Return labeled predictions and input status in a clear table, preserve input order, and provide a useful indication of applicability based on the method you evaluated. Reject invalid or unsupported inputs explicitly instead of returning misleading numbers. Demonstrate several chemically contrasting molecules plus invalid input; identify any demonstration molecule already present in training. Do not present familiar training examples as independent validation. Save the complete inference pipeline and verify that it can be reloaded in a fresh Python process to reproduce predictions without retraining.
