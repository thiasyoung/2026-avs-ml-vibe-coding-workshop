# 4. Boiling point: one-shot vibe-coding specification

**Demo purpose:** after showing the finished workflow, ask an LLM to create a comparable pipeline in one well-scoped request. This is a teaching prompt, not a claim that unreviewed generated code is scientifically valid.

Use the prompt below in Codex or another coding agent from the `workshop code` folder. Ask the agent to make a new output folder, never overwrite the prepared data, and explain every assumption it makes.


## One-shot prompt

> Build a reproducible molecular-regression pipeline in a new folder named `outputs/bp_vibe_coded`. My prepared data are `data/01_Structure_to_Boiling_Point/CompoundData.parquet`; use only `SMILES` as input and `BP` as the target. Inspect the schema and validate missing values, invalid SMILES, duplicates, units, and target distribution before modeling. Randomly split unique molecules into 80% training, 10% validation, and 10% test with seed 42. Use Chemprop’s directed message-passing neural network and do not leak targets into descriptors or preprocessing. Record training and validation loss every epoch, retain the epoch with the lowest validation loss, save the model and a manifest of commands/settings, predict every split, and report MAE, RMSE, and R². Produce measured-versus-predicted parity plots for all splits and a short `README.md` explaining the result and limitations of a random molecular split. Before writing code, give me a concise plan and identify missing dependencies. Then implement and run the plan.


## Review checklist

1. Did the generated code inspect and preserve the source data instead of silently changing it?
2. Is the split performed before any learned preprocessing, and are duplicate structures kept within one split?
3. Is model selection based on validation data, with the test set reserved for the final report?
4. Can another participant rerun the result from the manifest?
