# Workshop roadmap

Use a coding assistant to turn a scientific question into a working, tested
machine-learning workflow. The modules emphasize different parts of that process:

| Module | Your focus | Keep fixed while comparing |
| --- | --- | --- |
| Boiling point | Build a full molecular-prediction pipeline using the ten-step template. | Documented data source/units and independent evaluation splits. |
| IR functional groups | Examine spectra and test data-carpentry hypotheses. | Classifier settings and frozen validation/test data. |
| HTEM process emulator | Test models, architectures and hyperparameters. | Prepared features, transforms and grouped splits. |

## Before the session

Follow [SETUP_GUIDE.md](SETUP_GUIDE.md) and run
[Check your installation](00_Test_Installation.ipynb).
Download the whole repository so helpers and data stay with the notebooks.

**IR requires a separate download:** get **ir_raw_checkpoint.parquet** from
[Workshop data v1 → Assets](https://github.com/thiasyoung/2026-avs-ml-vibe-coding-workshop/releases/tag/workshop-data-v1),
not Source code, and save it at
`02_ir_functional_groups/outputs/nist_acquisition/ir_raw_checkpoint.parquet`.
BP has a packaged data fallback and HTEM includes compact snapshots/frozen features;
those cannot replace the native IR checkpoint. Full live acquisition is pre-work,
not a required workshop activity.

## 01 · Boiling point

- [00 — Get data](01_boiling_point/00_BP_Get_Data_NISTChemPy_v2.ipynb): acquire or cleanse NIST SMILES/BP data in kelvin.
- [01 — Ten steps in prompts](01_boiling_point/01_Ten_Steps_in_Prompts.ipynb): review the template.
- [02 — Vibe-coding workbook](01_boiling_point/02_BP_Vibe_Coding_Workbook.ipynb): adapt the opening prompt and generate your pipeline.

Optional [BP examples](01_boiling_point/examples/README.md) show Chemprop, frozen
MoLFormer and one-shot approaches. MoLFormer downloads its checkpoint on first use;
arrange internet access or a populated Hugging Face cache if you plan to run it.

## 02 · IR functional groups

- [00 — Get data](02_ir_functional_groups/00_IR_Get_Data_NISTChemPy_v2.ipynb): inspect/reuse native spectra, units, SMILES and labels.
- [01 — Data-carpentry workbook](02_ir_functional_groups/01_IR_Vibe_Coding_Data_Carpentry.ipynb): investigate a hypothesis and create uniform input files.
- [02 — Static training](02_ir_functional_groups/02_IR_Optimized_Training.ipynb): select that dataset and evaluate the fixed classifier.

Your workbook writes three split Parquets, a manifest and an audit under
`outputs/workshop_carpentry/`. The fixed output contract keeps them compatible with
02. Optional [IR examples](02_ir_functional_groups/examples/README.md) provide a
minimal baseline, alternative carpentry choices and a reference comparison.

## 03 · HTEM process emulator

- [00 — Get data](03_htem_conductivity/00_HTEM_Get_Data_NREL.ipynb): compact process, scalar-property and substrate-position data.
- [01 — Prepare data](03_htem_conductivity/01_HTEM_Process_Data_Carpentry.ipynb): supplied preparation of process/source/XY features and grouped splits.
- [02 — Model-experiment workbook](03_htem_conductivity/02_HTEM_Vibe_Coding_Workbook.ipynb): generate and analyze model experiments for local thickness/conductivity.

The goal is one prediction per substrate location, reconstructed into a map.
Measured film properties/spectra are not inputs. Optional
[HTEM examples](03_htem_conductivity/examples/README.md) illustrate several algorithms
and their reference comparison; they are not additional required session steps.


## Work in your own workbook

- In BP, use the ten-step template to guide your workflow. In IR and HTEM, edit
  the **EDITABLE** starting plan and blanks for your hypothesis, strategy and effort
  priorities; keep only their **FIXED input/output contracts** unchanged.
- Copy the entire prompt into your coding assistant with the repository open.
  A prompt-only notebook does not call an AI when you press Run All.
- Use `examples/` as references and leave supplied datasets/holdouts intact.
  Shared IR/HTEM helpers live in `lib/`; the setup cells find them automatically.
- Keep independent attempts in separate output folders. Review actual outputs,
  compare on validation data, evaluate fixed choices on test data and describe limitations.
- The reading website displays saved results with expandable code; it does not
  execute notebooks. Some saved models/run folders must be regenerated to rerun
  an optional comparison. You do not need to rerun every example to follow the workshop.
