# AVS workshop: three scientific-machine-learning demos

Each example lives in its own numbered folder. Notebook `00` is executable acquisition code for the source repository; run it during preparation and keep its output available.

Use the two-digit filename prefix for notebook order within each module. Lettered examples are alternatives to that numbered step, not additional main steps.

## Workshop conventions

- Retained Parquet files under each `data/` folder provide an offline/rehearsal fallback; the original course material is not required for this repository.
- Each stochastic action has a fixed seed, and evaluated models keep validation/test examples out of fitting and preprocessing.
- The IR trainer saves weights every 25 epochs, restores the lowest validation-loss model, logs TensorBoard events, selects thresholds on validation data only, and reports untouched-test metrics.
- The foundation-model notebook downloads a public MoLFormer checkpoint on its first run; rehearse it with internet access or a pre-populated Hugging Face cache.
