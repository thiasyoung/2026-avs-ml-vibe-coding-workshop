### Check the workshop kernel

Code Summary: Checks that the notebook is running from the workshop folder with Python 3.11 and the workshop virtual environment.


```python
from pathlib import Path
import sys

print(f'Python: {sys.version.split()[0]}')
print(f'Executable: {sys.executable}')
assert sys.version_info[:2] == (3, 11), 'This workshop requires Python 3.11.'
assert Path('.venv').exists(), 'Set up the workshop with UV from the repository root first.'
assert '.venv' in Path(sys.executable).parts, 'Select the workshop .venv as the notebook kernel in VS Code.'
print('OK  Python 3.11 in the workshop .venv')
```

### Verify required packages

Code Summary: Checks that the packages used throughout the workshop can be imported. This verifies installation without downloading data or training a model.


```python
from importlib import import_module

packages = {
    'Chemprop': 'chemprop', 'JupyterLab': 'jupyterlab', 'NISTChemPy': 'nistchempy',
    'NumPy': 'numpy', 'Pandas': 'pandas', 'PyTorch': 'torch', 'RDKit': 'rdkit',
    'scikit-learn': 'sklearn', 'TensorBoard': 'tensorboard', 'Transformers': 'transformers',
}
for label, module in packages.items():
    import_module(module)
    print(f'OK  {label}')
print('\nEnvironment ready.')
```
